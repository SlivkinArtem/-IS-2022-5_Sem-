### Задача 1

```python
def restore_set(s):
    result = []
    intervals = s.split(',')
    
    for interval in intervals:
        if '-' in interval:
            start, end = map(int, interval.split('-'))
            result.extend(range(start, end + 1))
        else:
            result.append(int(interval))
    
    return result

if __name__ == '__main__':
    restored_set = restore_set(input())
    print(' '.join(map(str, restored_set)))
```

асимптотика по времени O(k), где k — количество всех чисел

### Задача 2

```python
def restore_snow(journal):
    n = len(journal)
    result = [-1] * n  # Массив для хранения количества выпавшего снега в каждый день

    # Проверяем корректность данных и восстанавливаем дни с кляксами
    for i in range(1, n):
        # Если текущий день и предыдущий известны, проверяем, корректны ли данные
        if journal[i] != -1 and journal[i-1] != -1:
            if journal[i] <= journal[i-1]:  # Глубина снега должна увеличиваться
                return "NO"

    # Восстанавливаем количество выпавшего снега
    for i in range(n):
        if journal[i] == -1:
            # Если день с кляксой, определяем минимальное возможное значение
            prev_depth = journal[i-1] if i > 0 else 0
            next_depth = journal[i+1] if i < n-1 and journal[i+1] != -1 else 10**9

            # Восстанавливаем глубину снега в пределах возможного интервала
            journal[i] = prev_depth + 1
            if journal[i] >= next_depth:
                return "NO"
    
    # Теперь вычисляем количество выпавшего снега
    result[0] = journal[0]  # На первый день выпало столько же снега, какова была глубина
    for i in range(1, n):
        result[i] = journal[i] - journal[i-1]  # Разница глубин между днями = количество снега

    return "YES\n" + " ".join(map(str, result))

# Чтение данных
n = int(input())  # Количество дней
journal = list(map(int, input().split()))  # Журнал с данными о снеге

# Вызов функции и вывод результата
print(restore_snow(journal))
```

Время: O(n), так как мы дважды проходим по массиву длиной n.\
Память: O(n) для хранения результата и восстановления глубин снега.

### Задача 3

```python
from collections import Counter

def find_password(sequence, required_chars, k):
    required_count = Counter(required_chars)  # Требуемые символы и их количество
    current_count = Counter()  # Считаем символы в текущем окне
    n = len(sequence)
    
    left = 0  # Левый указатель окна
    best_password = ''  # Наилучшая найденная подстрока
    required_len = len(required_chars)  # Количество уникальных символов в наборе
    
    # Перемещаем правый указатель окна
    for right in range(n):
        current_char = sequence[right]
        
        # Если символ в наборе требований, увеличиваем его счетчик
        if current_char in required_count:
            current_count[current_char] += 1
        
        # Сжимаем окно слева, если в нем слишком много символов
        while (right - left + 1) > k:
            left_char = sequence[left]
            if left_char in current_count:
                current_count[left_char] -= 1
                if current_count[left_char] == 0:
                    del current_count[left_char]
            left += 1
        
        # Проверяем, содержит ли текущее окно все необходимые символы
        if all(current_count[char] >= required_count[char] for char in required_count):
            # Если текущее окно удовлетворяет условиям, проверяем длину и начало
            current_password = sequence[left:right + 1]
            if len(current_password) <= k and (len(current_password) > len(best_password) or left > best_password_start):
                best_password = current_password
                best_password_start = left
    
    return best_password if best_password else "-1"


# Пример использования
sequence = input().strip()
required_chars = input().strip()
k = int(input().strip())

# Вызов функции и вывод результата
print(find_password(sequence, required_chars, k))

            
```

Время: O(n), где n — длина последовательности. Мы проходим по строке один раз с помощью двух указателей.\
Память: O(m), где m — количество уникальных символов в строке required_chars

### Задача 4

```python
import math

# Функция для проверки, является ли число простым
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(math.sqrt(n)) + 1):
        if n % i == 0:
            return False
    return True

# Функция для подсчета количества делителей числа
def count_divisors(n):
    count = 0
    for i in range(1, int(math.sqrt(n)) + 1):
        if n % i == 0:
            count += 1
            if i != n // i:  # Если делители разные
                count += 1
    return count

# Основная функция
def count_composite_with_prime_divisors(l, r):
    count = 0
    
    for n in range(l, r + 1):
        if n == 1 or is_prime(n):  # Пропускаем единицу и простые числа
            continue
        
        divisors_count = count_divisors(n)
        if is_prime(divisors_count):  # Проверяем, является ли количество делителей простым
            count += 1
    
    return count

# Ввод
l, r = map(int, input().split())

# Вывод результата
print(count_composite_with_prime_divisors(l, r))
```

Время работы функции count_composite_with_prime_divisors: для каждого числа n из диапазона, проверка на простоту числа и подсчёт количества делителей занимает O(n). Если диапазон от l до r содержит d=r−l+1 чисел, то общая сложность будет O(d×r)

### Задача 5

```python
from collections import defaultdict

def time_to_seconds(hh, mm, ss):
    """Преобразуем время в формат (часы, минуты, секунды) в количество секунд с начала дня"""
    return int(hh) * 3600 + int(mm) * 60 + int(ss)

def parse_time(time_str):
    """Парсим строку времени формата hh:mm:ss и возвращаем количество секунд"""
    hh, mm, ss = time_str.split(':')
    return time_to_seconds(hh, mm, ss)

def main():
    # Чтение данных
    start_time = parse_time(input().strip())
    n = int(input().strip())
    
    # Словари для хранения информации о командах
    team_data = defaultdict(lambda: {"score": 0, "penalty": 0, "attempts": defaultdict(int)})
    
    for _ in range(n):
        line = input().strip().split()
        team_name = line[0][1:-1]  # Убираем кавычки
        event_time = parse_time(line[1])
        server_id = line[2]
        result = line[3]
        
        # Если запрос не "PING" (PONG), обрабатываем
        if result == "PONG":
            continue
        
        # Время в минутах с начала хакатона
        time_elapsed = (event_time - start_time) // 60
        if time_elapsed < 0:
            time_elapsed += 24 * 60  # Коррекция на переход через полночь

        # Обработка взлома сервера
        if result == "ACCESSED":
            # Добавляем очки и штраф за неудачные попытки
            team_data[team_name]["score"] += 1
            team_data[team_name]["penalty"] += team_data[team_name]["attempts"][server_id] * 20
            team_data[team_name]["penalty"] += time_elapsed
            
            # Очищаем информацию о неудачных попытках для этого сервера
            team_data[team_name]["attempts"][server_id] = 0
        else:
            # Неудачная попытка взлома (FORBIDEN или DENIED)
            team_data[team_name]["attempts"][server_id] += 1
    
    # Формируем список с результатами
    result_list = []
    for team_name, data in team_data.items():
        result_list.append((team_name, data["score"], data["penalty"]))
    
    # Сортировка по количеству взломов (по убыванию), затем по штрафам (по возрастанию), и по имени
    result_list.sort(key=lambda x: (-x[1], x[2], x[0]))
    
    # Формируем вывод
    rank = 1
    previous_score = -1
    previous_penalty = -1
    for i, (team_name, score, penalty) in enumerate(result_list):
        if i > 0 and (score < previous_score or penalty > previous_penalty):
            rank = i + 1
        previous_score = score
        previous_penalty = penalty
        print(f'{rank} "{team_name}" {score} {penalty}')

if __name__ == "__main__":
    main()
```

Время: O(nlogn). Основная сложность возникает на этапе сортировки результатов, что даёт сложность O(mlogm), где m — число уникальных команд (в худшем случае m=n).

### Задача 6

```python
from collections import deque, defaultdict

def find_min_completion_time(n, processes):
    # Время выполнения каждого процесса
    times = [0] * n
    # Граф зависимостей
    graph = defaultdict(list)
    # Количество входящих рёбер (зависимостей)
    indegree = [0] * n
    
    # Считывание входных данных
    for i in range(n):
        data = processes[i]
        times[i] = data[0]  # Время выполнения i-го процесса
        for dependency in data[1:]:
            graph[dependency - 1].append(i)  # Добавляем зависимость
            indegree[i] += 1  # Увеличиваем количество зависимостей
    
    # Массив для хранения максимального времени завершения каждого процесса
    completion_time = [0] * n
    
    # Очередь для топологической сортировки
    queue = deque()
    
    # Ищем все процессы, которые можно начать сразу (нет зависимостей)
    for i in range(n):
        if indegree[i] == 0:
            queue.append(i)
            completion_time[i] = times[i]
    
    # Топологическая сортировка
    while queue:
        process = queue.popleft()
        
        # Для всех процессов, зависящих от текущего
        for dependent in graph[process]:
            # Обновляем максимальное время завершения зависимых процессов
            completion_time[dependent] = max(completion_time[dependent], completion_time[process] + times[dependent])
            # Уменьшаем количество оставшихся зависимостей
            indegree[dependent] -= 1
            # Если больше нет зависимостей, добавляем в очередь
            if indegree[dependent] == 0:
                queue.append(dependent)
    
    # Ответ - максимальное время завершения всех процессов
    return max(completion_time)

if __name__ == "__main__":
    # Чтение данных через input()
    n = int(input().strip())  # Число процессов
    processes = []
    
    for _ in range(n):
        # Считываем строку данных, где первое число — время выполнения, а остальные — зависимости
        data = list(map(int, input().strip().split()))
        processes.append(data)
    
    print(find_min_completion_time(n, processes))  
```

Асимптотика: Построение графа и заполнение очереди — O(n), где n — количество процессов. Топологическая сортировка — O(n+m), где m — количество зависимостей. Таким образом, общая сложность алгоритма — O(n+m), что достаточно эффективно для n до 10^5 .
