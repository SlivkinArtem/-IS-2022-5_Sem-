```python
import re
def remove_bracketed_sections(text):
    # Удаляет все текстовые секции, заключенные в квадратные скобки, из текста.
    text = re.sub(r'\[.*?\]', '', text)
    # Создает таблицу перевода для замены определенных символов
    translation_table = str.maketrans({';': '', '"': '', ' ': ' '})
    # Применяет таблицу перевода к тексту, заменяя указанные символы на пробелы.
    return text.translate(translation_table)
```

```python
df['lyrics'] = df['lyrics'].apply(remove_bracketed_sections)
```

```python
df['lyrics'][129]
```

```python
def preprocess_lyrics(text):
    # Убираем тройные переводы строк
    text = re.sub(r'\n{2,}', '\n', text)
    text = re.sub(r'\(.*?\)', '', text)
    # Убираем лишние пробелы и пустые строки
    text = re.sub(r'\s*\n\s*', '\n', text)  # Убираем пробелы вокруг переносов строк
    # Убираем начальные и конечные пробелы
    text = text.strip()
    return text
```

```python
df['lyrics'] = df['lyrics'].apply(preprocess_lyrics)
df['lyrics'][129]
```

```python
all_songs = "\n\n".join(df['lyrics'].dropna())
with open("all_songs.txt", "w", encoding="utf-8") as f:
    f.write(all_songs)
```

```python
from transformers import GPT2LMHeadModel, Trainer, TrainingArguments
from datasets import Dataset
from transformers import GPT2Tokenizer

tokenizer = GPT2Tokenizer.from_pretrained("gpt2")

if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token  # Используем eos_token в качестве pad_token

# Сохранение обновленного токенизатора
tokenizer.save_pretrained('./tokenizer/')

# Загружаем модель
model = GPT2LMHeadModel.from_pretrained("gpt2")

# Создание датасета
dataset = Dataset.from_dict({"text": [all_songs]})

def tokenize_function(examples):
    tokens = tokenizer(
        examples["text"],
        truncation=True,
        padding="max_length",
        max_length=512,
    )
    # Добавляем метки, которые совпадают с input_ids
    tokens["labels"] = tokens["input_ids"].copy()
    return tokens


tokenized_dataset = dataset.map(tokenize_function, batched=True)
```

```python
# from transformers import Trainer, TrainingArguments

training_args = TrainingArguments(
    output_dir="./results",
    num_train_epochs=3,
    per_device_train_batch_size=2,
    save_steps=10_000,
    save_total_limit=2,
    prediction_loss_only=False,  # Включаем расчёт потерь
    logging_dir='./logs',
    report_to=[],
)

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_dataset,
)
```

```python
trainer.train()
# Сохранение обученной модели
trainer.save_model("./results")  # Сохраняет модель и её конфигурацию
tokenizer.save_pretrained("./results")  # Сохраняет токенизатор
```

```python
model = GPT2LMHeadModel.from_pretrained("./results")
tokenizer = GPT2Tokenizer.from_pretrained("./tokenizer")
model.eval()
```

```python
from transformers import TextGenerationPipeline

generator = TextGenerationPipeline(model=model, tokenizer=tokenizer)
def generate_text(prompt, max_length=100, num_return_sequences=1):
  results = generator(
        prompt,
        max_length=max_length,
        num_return_sequences=num_return_sequences,
        do_sample=True,  # Включает семплирование для более разнообразных результатов
        temperature=0.7,  # Управляет "творческой свободой" генерации (меньше - более детерминированно)
        top_k=50,  # Учитывает только топ-50 вероятных слов
        top_p=0.95,  # Nucleus Sampling, учитывает слова с вероятностью до 95%
    )
  return [result["generated_text"] for result in results]
```

```python
# Пример начального текста
prompt = "Курить и сдохнуть, схвативши оголённый провод как дружеский локоть"

# Генерация текста
generated_texts = generate_text(prompt, max_length=200, num_return_sequences=1)

# Печать результатов
for idx, text in enumerate(generated_texts, 1):
    print(f"Сгенерированный текст {idx}:\n{text}\n")
```
